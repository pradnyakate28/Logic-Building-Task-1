```python
# Task 1: Check Login

username = "admin"
password = "1234"

input_user = "admin"
input_pass = "1234"

if input_user == username and input_pass == password:
    print("Login Successful")
else:
    print("Invalid Credentials")
```

```python
# Task 2: Count Pass and Fail Students

marks = [45, 78, 90, 33, 60]

pass_students = 0
fail_students = 0

for m in marks:
    if m >= 50:
        pass_students = pass_students + 1
    else:
        fail_students = fail_students + 1

print("Total Students:", len(marks))
print("Pass Students:", pass_students)
print("Fail Students:", fail_students)
```

```python
# Task 3: Clean Names

names = [" Alice ", "bob", " CHARLIE "]

clean_names = []

for name in names:
    name = name.strip()      # remove spaces
    name = name.lower()      # convert to lowercase
    clean_names.append(name)

print("Cleaned List:", clean_names)
```

```python
# Task 4: Check Message Length

messages = ["Hi", "Welcome to the platform", "OK"]

for msg in messages:
    length = len(msg)
    print("Message:", msg)
    print("Length:", length)

    if length > 10:
        print("This message is longer than 10 characters.")

    print("-------------------")
```

```python
# Task 5: Count ERROR in Logs

logs = ["INFO", "ERROR", "WARNING", "ERROR"]

error_count = 0

for log in logs:
    if log == "ERROR":
        error_count = error_count + 1

print("Total ERROR entries:", error_count)
```
